## Dataset Structure & Segmentation Strategy

**Dataset Structure:**
- **Transactional data**: Each row represents one product in an order/invoice
- Multiple rows can have the same InvoiceNo (multi-item orders)
- Each invoice belongs to one CustomerID
- Data spans multiple countries and time periods

**Key for Customer Segmentation:**
- **CustomerID**: Primary key for aggregating customer-level features
- **Note**: Some transactions have missing CustomerID (guest purchases) - we'll need to filter these

**Potential Features for Customer Segmentation:**

1. **RFM Metrics** (Classic segmentation approach):
   - **Recency**: Days since last purchase
   - **Frequency**: Number of orders/invoices
   - **Monetary**: Total revenue generated

2. **Behavioral Features**:
   - Average order value
   - Average basket size (items per order)
   - Purchase frequency (orders per month)
   - Product diversity (unique products purchased)
   - Preferred shopping time (hour/day patterns)

3. **Product Preferences**:
   - Most purchased product categories
   - Price sensitivity (avg unit price)
   - Return rate (negative quantities indicate returns)

4. **Geographic & Temporal**:
   - Country
   - Time span as customer
   - Consistency of purchases (std dev of order intervals)

# Feature Engineering for Customer Segmentation

## Data Cleaning & Filtering

```python
# Create a clean dataframe for feature engineering
# Remove rows with missing CustomerID
df_clean = df[df['CustomerID'].notna()].copy()

# Remove cancelled orders (InvoiceNo starting with 'C')
df_clean = df_clean[~df_clean['InvoiceNo'].astype(str).str.startswith('C')]

# Calculate revenue
df_clean['Revenue'] = df_clean['Quantity'] * df_clean['UnitPrice']

# Get reference date (one day after last transaction for recency calculation)
reference_date = df_clean['InvoiceDate'].max() + pd.Timedelta(days=1)

print("Data Cleaning Summary:")
print(f"Original transactions: {len(df)}")
print(f"After removing missing CustomerID: {len(df[df['CustomerID'].notna()])}")
print(f"After removing cancellations: {len(df_clean)}")
print(f"Unique customers for analysis: {df_clean['CustomerID'].nunique()}")
print(f"\nReference date for recency: {reference_date}")
```

## RFM Features (Recency, Frequency, Monetary)

```python
# Calculate RFM features
rfm = df_clean.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (reference_date - x.max()).days,  # Recency
    'InvoiceNo': 'nunique',  # Frequency
    'Revenue': 'sum'  # Monetary
}).reset_index()

rfm.columns = ['CustomerID', 'Recency', 'Frequency', 'Monetary']

print("RFM Features:")
print(rfm.head(10))
print(f"\nRFM Statistics:")
print(rfm[['Recency', 'Frequency', 'Monetary']].describe())
```

## Purchase Behavior Features

```python
# Calculate Purchase Behavior features
behavior = df_clean.groupby('CustomerID').agg({
    'Quantity': 'sum',  # Total items purchased
    'StockCode': 'nunique',  # Unique products
    'InvoiceDate': ['min', 'max']  # Date range
}).reset_index()

behavior.columns = ['CustomerID', 'TotalItems', 'UniqueProducts', 'FirstPurchase', 'LastPurchase']

# Calculate purchase span (days as customer)
behavior['PurchaseSpan'] = (behavior['LastPurchase'] - behavior['FirstPurchase']).dt.days

# Calculate average basket size and average order value per customer
basket_stats = df_clean.groupby(['CustomerID', 'InvoiceNo']).agg({
    'Quantity': 'sum',
    'Revenue': 'sum'
}).reset_index()

basket_avg = basket_stats.groupby('CustomerID').agg({
    'Quantity': 'mean',
    'Revenue': 'mean'
}).reset_index()

basket_avg.columns = ['CustomerID', 'AvgBasketSize', 'AvgOrderValue']

# Merge with behavior
behavior = behavior.merge(basket_avg, on='CustomerID')

# Drop date columns (already captured in Recency and PurchaseSpan)
behavior = behavior.drop(['FirstPurchase', 'LastPurchase'], axis=1)

print("Purchase Behavior Features:")
print(behavior.head(10))
print(f"\nBehavior Statistics:")
print(behavior.drop('CustomerID', axis=1).describe())
```

## Pattern Features (Temporal & Shopping Patterns)

```python
# Calculate Pattern features
# Get order dates per customer
order_dates = df_clean.groupby(['CustomerID', 'InvoiceNo'])['InvoiceDate'].min().reset_index()
order_dates = order_dates.sort_values(['CustomerID', 'InvoiceDate'])

# Calculate days between orders
def calculate_order_intervals(group):
    if len(group) > 1:
        intervals = group['InvoiceDate'].diff().dt.days.dropna()
        return pd.Series({
            'AvgDaysBetweenOrders': intervals.mean() if len(intervals) > 0 else 0,
            'StdDaysBetweenOrders': intervals.std() if len(intervals) > 0 else 0
        })
    else:
        return pd.Series({
            'AvgDaysBetweenOrders': 0,
            'StdDaysBetweenOrders': 0
        })

intervals_df = order_dates.groupby('CustomerID').apply(calculate_order_intervals).reset_index()

# Get preferred day of week and hour
time_prefs = df_clean.groupby('CustomerID').agg({
    'DayOfWeek': lambda x: x.mode()[0] if len(x.mode()) > 0 else x.iloc[0],
    'Hour': lambda x: x.mode()[0] if len(x.mode()) > 0 else x.iloc[0]
}).reset_index()

time_prefs.columns = ['CustomerID', 'PreferredDayOfWeek', 'PreferredHour']

# Merge pattern features
patterns = intervals_df.merge(time_prefs, on='CustomerID')

print("Pattern Features:")
print(patterns.head(10))
print(f"\nPattern Statistics:")
print(patterns.drop('CustomerID', axis=1).describe())
```

## Product Preference Features

```python
# Calculate Product Preference features
product_prefs = df_clean.groupby('CustomerID').agg({
    'UnitPrice': 'mean',  # Average unit price (price sensitivity)
    'Quantity': ['sum', lambda x: (x < 0).sum()]  # Total quantity and returns count
}).reset_index()

product_prefs.columns = ['CustomerID', 'AvgUnitPrice', 'TotalQuantity', 'ReturnCount']

# Calculate return rate (% of transactions that are returns)
transaction_counts = df_clean.groupby('CustomerID').size().reset_index(name='TotalTransactions')
product_prefs = product_prefs.merge(transaction_counts, on='CustomerID')
product_prefs['ReturnRate'] = product_prefs['ReturnCount'] / product_prefs['TotalTransactions']

# Calculate exploration tendency (unique products / total items)
# Already have TotalItems and UniqueProducts in behavior df, will merge later
product_prefs = product_prefs.drop(['TotalQuantity', 'ReturnCount', 'TotalTransactions'], axis=1)

print("Product Preference Features:")
print(product_prefs.head(10))
print(f"\nProduct Preference Statistics:")
print(product_prefs.drop('CustomerID', axis=1).describe())
```

## Geographic Features

```python
# Get primary country per customer (most frequent)
geographic = df_clean.groupby('CustomerID')['Country'].agg(lambda x: x.mode()[0] if len(x.mode()) > 0 else x.iloc[0]).reset_index()
geographic.columns = ['CustomerID', 'Country']

print("Geographic Features:")
print(geographic.head(10))
print(f"\nCountry Distribution:")
print(geographic['Country'].value_counts().head(10))
```

## Merge All Features

```python
# Merge all feature sets
features = rfm.copy()
features = features.merge(behavior, on='CustomerID', how='left')
features = features.merge(patterns, on='CustomerID', how='left')
features = features.merge(product_prefs, on='CustomerID', how='left')
features = features.merge(geographic, on='CustomerID', how='left')

# Calculate exploration tendency (unique products / total items)
features['ExplorationTendency'] = features['UniqueProducts'] / features['TotalItems']

# Fill NaN values in interval features (for customers with only 1 order)
features['AvgDaysBetweenOrders'].fillna(0, inplace=True)
features['StdDaysBetweenOrders'].fillna(0, inplace=True)

print("Merged Feature Matrix:")
print(features.head(10))
print(f"\nFeature Matrix Shape: {features.shape}")
print(f"\nColumns: {features.columns.tolist()}")
print(f"\nMissing values:")
print(features.isnull().sum())
```

## Encode Country & Standardize Features

```python
# One-hot encode Country (keep top 10 countries, group rest as 'Other')
top_countries = features['Country'].value_counts().head(10).index
features['Country_Grouped'] = features['Country'].apply(lambda x: x if x in top_countries else 'Other')

# One-hot encode
country_dummies = pd.get_dummies(features['Country_Grouped'], prefix='Country')
features = pd.concat([features, country_dummies], axis=1)

# Drop original Country columns
features = features.drop(['Country', 'Country_Grouped'], axis=1)

print("Features after encoding Country:")
print(f"Shape: {features.shape}")
print(f"Columns: {features.columns.tolist()}")

# Separate CustomerID from features
customer_ids = features['CustomerID'].copy()
feature_cols = [col for col in features.columns if col != 'CustomerID']

# Standardize features
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
features_scaled = scaler.fit_transform(features[feature_cols])

# Create final dataframe
features_final = pd.DataFrame(features_scaled, columns=feature_cols)
features_final.insert(0, 'CustomerID', customer_ids.values)

print("\n✓ Feature engineering complete!")
print(f"Final feature matrix shape: {features_final.shape}")
print(f"Features: {len(feature_cols)}")
print(f"\nFirst few rows:")
print(features_final.head())
```
