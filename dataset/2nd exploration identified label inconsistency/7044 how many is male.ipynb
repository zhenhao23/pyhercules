# Statistics for Cluster 1526

This notebook joins the two datasets and shows separate statistics for customers in L1_cluster_id = 1526:
- How many are male
- How many have a partner
- How many are senior citizens

```python
import pandas as pd
```

## Load and Join Datasets

```python
# Load both datasets
telco_df = pd.read_csv('../Telco customer churn IBM dataset/Telco_customer_churn.csv')
retention_df = pd.read_csv('Retention strategy for newly subscribed customers k+1.csv')

# Reset index and join by row number
telco_df_reset = telco_df.reset_index(drop=True)
retention_df_reset = retention_df.reset_index(drop=True)

# Join the datasets
joined_df = pd.concat([telco_df_reset, retention_df_reset], axis=1)

print(f"Joined dataset shape: {joined_df.shape}")
print(f"Columns in joined dataset: {joined_df.columns.tolist()}")
```

## Statistics for Cluster 1526

```python
# Filter by L1_cluster_id = 1526
cluster_1526 = joined_df[joined_df['L1_cluster_id'] == 7045]

total = len(cluster_1526)
print(f"Total customers in cluster 1526: {total}")
print(f"\n{'='*60}")

# Calculate percentage of Males
male_count = len(cluster_1526[cluster_1526['Gender'] == 'Male'])
male_percentage = (male_count / total * 100) if total > 0 else 0
print(f"Percentage of Males: {male_percentage:.2f}% ({male_count}/{total})")

# Calculate percentage with Partner = Yes
partner_yes_count = len(cluster_1526[cluster_1526['Partner'] == 'Yes'])
partner_percentage = (partner_yes_count / total * 100) if total > 0 else 0
print(f"Percentage with Partner = Yes: {partner_percentage:.2f}% ({partner_yes_count}/{total})")

# Calculate percentage of Senior Citizens
senior_citizen_count = len(cluster_1526[cluster_1526['Senior Citizen'] == 'Yes'])
senior_percentage = (senior_citizen_count / total * 100) if total > 0 else 0
print(f"Percentage of Senior Citizens: {senior_percentage:.2f}% ({senior_citizen_count}/{total})")

print(f"{'='*60}")
```

## Deviation from Global Averages

Compare cluster 7045 statistics with the overall dataset averages:

```python
# Calculate global averages from the original dataset
global_male_pct = (len(telco_df[telco_df['Gender'] == 'Male']) / len(telco_df) * 100)
global_partner_pct = (len(telco_df[telco_df['Partner'] == 'Yes']) / len(telco_df) * 100)
global_senior_pct = (len(telco_df[telco_df['Senior Citizen'] == 'Yes']) / len(telco_df) * 100)

print(f"{'='*60}")
print(f"COMPARISON: Cluster 7045 vs Global Dataset")
print(f"{'='*60}\n")

# Male percentage comparison
male_deviation = male_percentage - global_male_pct
print(f"Male Percentage:")
print(f"  Global Average: {global_male_pct:.2f}%")
print(f"  Cluster 7045:   {male_percentage:.2f}%")
print(f"  Deviation:      {male_deviation:+.2f} percentage points")
print(f"  ({abs(male_deviation):.2f} percentage points {'higher' if male_deviation > 0 else 'lower'} than expected)\n")

# Partner percentage comparison
partner_deviation = partner_percentage - global_partner_pct
print(f"Partner = Yes Percentage:")
print(f"  Global Average: {global_partner_pct:.2f}%")
print(f"  Cluster 7045:   {partner_percentage:.2f}%")
print(f"  Deviation:      {partner_deviation:+.2f} percentage points")
print(f"  ({abs(partner_deviation):.2f} percentage points {'higher' if partner_deviation > 0 else 'lower'} than expected)\n")

# Senior Citizen percentage comparison
senior_deviation = senior_percentage - global_senior_pct
print(f"Senior Citizen Percentage:")
print(f"  Global Average: {global_senior_pct:.2f}%")
print(f"  Cluster 7045:   {senior_percentage:.2f}%")
print(f"  Deviation:      {senior_deviation:+.2f} percentage points")
print(f"  ({abs(senior_deviation):.2f} percentage points {'higher' if senior_deviation > 0 else 'lower'} than expected)\n")

print(f"{'='*60}")
print(f"SUMMARY:")
print(f"{'='*60}")
print(f"This cluster is:")
print(f"  • {abs(male_deviation):.2f}pp {'MORE' if male_deviation > 0 else 'LESS'} male than average")
print(f"  • {abs(partner_deviation):.2f}pp {'MORE' if partner_deviation > 0 else 'LESS'} likely to have partners")
print(f"  • {abs(senior_deviation):.2f}pp {'MORE' if senior_deviation > 0 else 'LESS'} likely to be senior citizens")
```

## Top 5 Features with Largest Deviation

Find the top 5 features where cluster 7045 deviates most from global averages:

```python
# Calculate deviations for numerical features
numerical_deviations = []

# Get all numerical columns from telco_df
numerical_columns = telco_df.select_dtypes(include=['int64', 'float64']).columns.tolist()

# Remove columns that are identifiers or geographic data
exclude_numeric_cols = ['Count', 'Zip Code', 'Latitude', 'Longitude']
numerical_columns = [col for col in numerical_columns if col not in exclude_numeric_cols]

# Convert Total Charges to numeric if it's stored as object/string
if 'Total Charges' in telco_df.columns and telco_df['Total Charges'].dtype == 'object':
    telco_df['Total Charges'] = pd.to_numeric(telco_df['Total Charges'], errors='coerce')
    cluster_1526['Total Charges'] = pd.to_numeric(cluster_1526['Total Charges'], errors='coerce')
    if 'Total Charges' not in numerical_columns:
        numerical_columns.append('Total Charges')

print(f"Analyzing numerical columns: {numerical_columns}\n")

# Calculate statistics for each numerical feature
for col in numerical_columns:
    # Global statistics
    global_mean = telco_df[col].mean()
    global_median = telco_df[col].median()
    global_std = telco_df[col].std()
    
    # Cluster statistics
    cluster_mean = cluster_1526[col].mean() if total > 0 else 0
    cluster_median = cluster_1526[col].median() if total > 0 else 0
    cluster_std = cluster_1526[col].std() if total > 0 else 0
    
    # Calculate deviations
    mean_deviation = cluster_mean - global_mean
    median_deviation = cluster_median - global_median
    
    # Calculate percentage deviation for mean
    mean_pct_deviation = (mean_deviation / global_mean * 100) if global_mean != 0 else 0
    
    numerical_deviations.append({
        'feature': col,
        'global_mean': global_mean,
        'cluster_mean': cluster_mean,
        'mean_deviation': mean_deviation,
        'mean_pct_deviation': mean_pct_deviation,
        'global_median': global_median,
        'cluster_median': cluster_median,
        'median_deviation': median_deviation,
        'abs_mean_deviation': abs(mean_deviation)
    })

# Calculate deviations for categorical features
categorical_deviations = []

# Get all categorical columns from telco_df
categorical_columns = telco_df.select_dtypes(include=['object']).columns.tolist()

# Remove columns that are identifiers, geographic data, or not useful for comparison
exclude_cols = ['CustomerID', 'Customer ID', 'Count', 'Country', 'State', 'City', 'Zip Code', 'Lat Long', 'Latitude', 'Longitude', 'Total Charges']
categorical_columns = [col for col in categorical_columns if col not in exclude_cols]

print(f"Analyzing categorical columns: {categorical_columns}\n")

# Calculate deviation for each categorical feature
for col in categorical_columns:
    # Get unique values for this column
    unique_values = telco_df[col].dropna().unique()
    
    for value in unique_values:
        # Calculate global percentage
        global_pct = (len(telco_df[telco_df[col] == value]) / len(telco_df) * 100)
        
        # Calculate cluster percentage
        cluster_pct = (len(cluster_1526[cluster_1526[col] == value]) / total * 100) if total > 0 else 0
        
        # Calculate deviation
        deviation = cluster_pct - global_pct
        
        categorical_deviations.append({
            'feature': f"{col} = {value}",
            'global_pct': global_pct,
            'cluster_pct': cluster_pct,
            'deviation': deviation,
            'abs_deviation': abs(deviation)
        })

# Display top 5 categorical deviations
top_5_categorical = sorted(categorical_deviations, key=lambda x: x['abs_deviation'], reverse=True)[:5]

# Display top numerical deviations
top_numerical = sorted(numerical_deviations, key=lambda x: abs(x['mean_pct_deviation']), reverse=True)

print(f"{'='*70}")
print(f"TOP NUMERICAL FEATURES WITH LARGEST DEVIATION: Cluster 7045 vs Global")
print(f"{'='*70}\n")

for i, item in enumerate(top_numerical, 1):
    print(f"{i}. {item['feature']}")
    print(f"   Global Mean:     {item['global_mean']:,.2f}")
    print(f"   Cluster Mean:    {item['cluster_mean']:,.2f}")
    print(f"   Deviation:       {item['mean_deviation']:+,.2f} ({item['mean_pct_deviation']:+.2f}%)")
    print(f"   Global Median:   {item['global_median']:,.2f}")
    print(f"   Cluster Median:  {item['cluster_median']:,.2f}")
    print()

print(f"{'='*70}")
print(f"TOP 5 CATEGORICAL FEATURES WITH LARGEST DEVIATION")
print(f"{'='*70}\n")

for i, item in enumerate(top_5_categorical, 1):
    print(f"{i}. {item['feature']}")
    print(f"   Global Average:  {item['global_pct']:6.2f}%")
    print(f"   Cluster 7045:    {item['cluster_pct']:6.2f}%")
    print(f"   Deviation:       {item['deviation']:+6.2f} percentage points")
    print(f"   ({item['abs_deviation']:.2f}pp {'higher' if item['deviation'] > 0 else 'lower'} than expected)")
    print()

print(f"{'='*70}")
print(f"KEY INSIGHTS:")
print(f"{'='*70}")
print("The features above show where this cluster differs most significantly")
print("from the overall customer population.")
```

```python
# Filter categorical data for LLM input
# Keep meaningful features and remove redundant "No" values

def should_keep_categorical_feature(feature_name):
    """
    Determine if a categorical feature should be kept for LLM input.
    Keep: "No internet service", "No phone service", multi-category features, Gender
    Remove: Any feature containing the word "No" (except "No internet service" and "No phone service")
    """
    # Always keep these meaningful "No" values
    if 'No internet service' in feature_name or 'No phone service' in feature_name:
        return True
    
    # Remove any feature containing "No" FIRST before checking other categories
    # This catches "Paperless = No", "Senior Citizen = No", "Churn Reason = Don't know", etc.
    if 'No' in feature_name:
        return False
    
    # Keep all multi-category features (Contract, Payment Method, Churn Reason)
    if any(keyword in feature_name for keyword in ['Contract =', 'Payment Method =', 'Churn Reason =']):
        return True
    
    # Keep both Gender values (not binary yes/no)
    if 'Gender =' in feature_name:
        return True
    
    # Keep everything else (Yes values, multi-category values)
    return True

# Filter categorical deviations
categorical_df_filtered = categorical_df_sorted[
    categorical_df_sorted['Feature'].apply(should_keep_categorical_feature)
].reset_index(drop=True)

# Combine numerical and categorical for LLM
# Add a "Type" column to distinguish between numerical and categorical
numerical_for_llm = numerical_df_sorted[['Feature', 'Global Mean', 'Cluster Mean', 'Mean % Deviation']].copy()
numerical_for_llm['Type'] = 'Numerical'
numerical_for_llm['Abs Deviation'] = numerical_for_llm['Mean % Deviation'].abs()
numerical_for_llm['Deviation Description'] = numerical_for_llm.apply(
    lambda row: f"Mean: {row['Cluster Mean']:.2f} vs Global: {row['Global Mean']:.2f} ({row['Mean % Deviation']:+.2f}%)",
    axis=1
)

categorical_for_llm = categorical_df_filtered[['Feature', 'Global %', 'Cluster 7045 %', 'Deviation (pp)', 'Absolute Deviation (pp)']].copy()
categorical_for_llm['Type'] = 'Categorical'
categorical_for_llm['Abs Deviation'] = categorical_for_llm['Absolute Deviation (pp)']
categorical_for_llm['Deviation Description'] = categorical_for_llm.apply(
    lambda row: f"Cluster: {row['Cluster 7045 %']:.2f}% vs Global: {row['Global %']:.2f}% ({row['Deviation (pp)']:+.2f}pp)",
    axis=1
)

# Create unified LLM input file with consistent columns
llm_input = pd.DataFrame({
    'Feature': list(numerical_for_llm['Feature']) + list(categorical_for_llm['Feature']),
    'Type': list(numerical_for_llm['Type']) + list(categorical_for_llm['Type']),
    'Abs Deviation': list(numerical_for_llm['Abs Deviation']) + list(categorical_for_llm['Abs Deviation']),
    'Deviation Description': list(numerical_for_llm['Deviation Description']) + list(categorical_for_llm['Deviation Description'])
})

# Sort by absolute deviation (largest first)
llm_input_sorted = llm_input.sort_values('Abs Deviation', ascending=False).reset_index(drop=True)

# Save filtered datasets
categorical_filtered_output = 'cluster_7045_categorical_filtered.csv'
llm_input_output = 'cluster_7045_llm_input.csv'

categorical_df_filtered.to_csv(categorical_filtered_output, index=False, float_format='%.2f')
llm_input_sorted.to_csv(llm_input_output, index=False, float_format='%.2f')

print(f"✓ Saved filtered categorical deviations to: {categorical_filtered_output}")
print(f"  Original categorical features: {len(categorical_df_sorted)}")
print(f"  Filtered categorical features: {len(categorical_df_filtered)}")
print(f"  Removed redundant features: {len(categorical_df_sorted) - len(categorical_df_filtered)}")

print(f"\n✓ Saved LLM-optimized input to: {llm_input_output}")
print(f"  Total features for LLM: {len(llm_input_sorted)}")
print(f"  - Numerical: {len(numerical_for_llm)}")
print(f"  - Categorical: {len(categorical_for_llm)}")
print(f"  Sorted by: Largest absolute deviation first")

print(f"\n{'='*70}")
print("FILTERED CATEGORICAL FEATURES (Top 10):")
print(f"{'='*70}")
print(categorical_df_filtered[['Feature', 'Cluster 7045 %', 'Global %', 'Deviation (pp)']].head(10).to_string(index=False))

print(f"\n{'='*70}")
print("LLM INPUT PREVIEW (Top 15 by Absolute Deviation):")
print(f"{'='*70}")
print(llm_input_sorted[['Feature', 'Type', 'Abs Deviation', 'Deviation Description']].head(15).to_string(index=False))
```
