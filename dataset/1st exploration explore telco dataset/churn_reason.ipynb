# Extract Unique Churn Reasons from Telco Dataset

This notebook reads the Telco customer churn dataset and extracts all unique values from the "Churn Reason" column.

```python
# Import required libraries
import pandas as pd
import os
```

```python
# Read the Excel file
file_path = r'C:\Users\weezh\OneDrive\Desktop\pyhercules\dataset\Telco customer churn IBM dataset\Telco_customer_churn.xlsx'
df = pd.read_excel(file_path)

# Display basic information about the dataset
print(f"Dataset shape: {df.shape}")
print(f"\nColumn names:\n{df.columns.tolist()}")
print(f"\nFirst few rows:")
df.head()
```

```python
# Extract unique values from 'Churn Reason' column
# Handle potential null values
unique_churn_reasons = df['Churn Reason'].dropna().unique()

print(f"Number of unique churn reasons: {len(unique_churn_reasons)}")
print(f"\nUnique Churn Reasons:")
for i, reason in enumerate(unique_churn_reasons, 1):
    print(f"{i}. {reason}")
```

```python
# Create a DataFrame with unique churn reasons
unique_reasons_df = pd.DataFrame({
    'Churn Reason': sorted(unique_churn_reasons)
})

# Also add count of each reason
reason_counts = df['Churn Reason'].value_counts()
unique_reasons_df['Count'] = unique_reasons_df['Churn Reason'].map(reason_counts)

print("Preview of the output:")
print(unique_reasons_df)
```

```python
# Export to CSV file
output_path = r'C:\Users\weezh\OneDrive\Desktop\pyhercules\dataset\unique_churn_reasons.csv'
unique_reasons_df.to_csv(output_path, index=False)

print(f"✓ Unique churn reasons exported successfully to:")
print(f"  {output_path}")
print(f"\nTotal unique reasons: {len(unique_reasons_df)}")
```
