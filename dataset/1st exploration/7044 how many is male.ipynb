# Count Male Customers in Cluster 7044

This notebook joins the two datasets and counts how many customers in L1_cluster_id = 7044 are male.

```python
import pandas as pd
```

## Load and Join Datasets

```python
# Load both datasets
telco_df = pd.read_csv('Telco customer churn IBM dataset/Telco_customer_churn.csv')
retention_df = pd.read_csv('Retention strategy for newly subscribed customers k+1.csv')

# Reset index and join by row number
telco_df_reset = telco_df.reset_index(drop=True)
retention_df_reset = retention_df.reset_index(drop=True)

# Join the datasets
joined_df = pd.concat([telco_df_reset, retention_df_reset], axis=1)

print(f"Joined dataset shape: {joined_df.shape}")
print(f"Columns in joined dataset: {joined_df.columns.tolist()}")
```

## Filter by Cluster 7044 and Count Male Customers

```python
# Filter by L1_cluster_id = 7044
cluster_7044 = joined_df[joined_df['L1_cluster_id'] == 7044]

print(f"Total customers in cluster 7044: {len(cluster_7044)}")
print(f"\nGender distribution in cluster 7044:")
print(cluster_7044['Gender'].value_counts())

# Count male customers
male_count = len(cluster_7044[cluster_7044['Gender'] == 'Male'])
female_count = len(cluster_7044[cluster_7044['Gender'] == 'Female'])

print(f"\n{'='*50}")
print(f"Number of MALE customers in cluster 7044: {male_count}")
print(f"Number of FEMALE customers in cluster 7044: {female_count}")
print(f"Percentage of males: {male_count/len(cluster_7044)*100:.2f}%")
print(f"{'='*50}")
```

## View Sample Data of Male Customers in Cluster 7044

```python
# Filter for male customers in cluster 7044
male_cluster_7044 = cluster_7044[cluster_7044['Gender'] == 'Male']

print(f"Sample of male customers in cluster 7044:")
print(male_cluster_7044[['Gender', 'L1_cluster_id', 'Churn Reason']].head(10))
```

## Analyze Female Customers with Partner Status in Cluster 7044

```python
# Filter for female customers in cluster 7044
female_cluster_7044 = cluster_7044[cluster_7044['Gender'] == 'Female']

print(f"Total FEMALE customers in cluster 7044: {len(female_cluster_7044)}")
print(f"\nPartner status distribution for females in cluster 7044:")
print(female_cluster_7044['Partner'].value_counts())

# Count female customers with Partner=Yes
female_with_partner = len(female_cluster_7044[female_cluster_7044['Partner'] == 'Yes'])
female_without_partner = len(female_cluster_7044[female_cluster_7044['Partner'] == 'No'])

print(f"\n{'='*50}")
print(f"Number of FEMALE customers with Partner=Yes: {female_with_partner}")
print(f"Number of FEMALE customers with Partner=No: {female_without_partner}")
print(f"Percentage with partner: {female_with_partner/len(female_cluster_7044)*100:.2f}%")
print(f"{'='*50}")
```
