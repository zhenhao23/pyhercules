# Join Two Datasets and Filter by Cluster

This notebook joins two CSV files by row number and performs filtering operations.

```python
import pandas as pd
import os
```

## Step 1: Load the datasets

```python
# Load the two datasets
telco_df = pd.read_csv('Telco customer churn IBM dataset/Telco_customer_churn.csv')
retention_df = pd.read_csv('Retention strategy for newly subscribed customers k+1.csv')

print(f"Telco dataset shape: {telco_df.shape}")
print(f"Retention dataset shape: {retention_df.shape}")
print(f"\nFirst few rows of Telco dataset:")
print(telco_df.head())
print(f"\nFirst few rows of Retention dataset:")
print(retention_df.head())
```

## Step 2: Join datasets by row number

```python
# Reset index to ensure row numbers align, then join by index
telco_df_reset = telco_df.reset_index(drop=True)
retention_df_reset = retention_df.reset_index(drop=True)

# Join the datasets by row number (index)
joined_df = pd.concat([telco_df_reset, retention_df_reset], axis=1)

print(f"Joined dataset shape: {joined_df.shape}")
print(f"\nFirst few rows of joined dataset:")
print(joined_df.head())

# Save the joined dataset to a new CSV file
joined_df.to_csv('joined_telco_retention.csv', index=False)
print(f"\nJoined dataset saved to 'joined_telco_retention.csv'")
```

## Step 3: Filter by L1_cluster_id = 7044

```python
# Filter the joined dataset by L1_cluster_id = 7043
filtered_df = joined_df[joined_df['L1_cluster_id'] == 7043]

print(f"Filtered dataset shape (L1_cluster_id = 7043): {filtered_df.shape}")
print(f"\nFirst few rows of filtered dataset:")
print(filtered_df.head())
```

## Step 4: Find unique Churn Reasons and save to CSV

```python
# Find unique values in Churn Reason column from the filtered dataset
unique_churn_reasons = filtered_df['Churn Reason'].value_counts().reset_index()
unique_churn_reasons.columns = ['Churn Reason', 'Count']

print(f"Unique Churn Reasons for L1_cluster_id = 7043:")
print(unique_churn_reasons)

# Save unique churn reasons to a new CSV file
unique_churn_reasons.to_csv('unique_churn_reasons_cluster_7043.csv', index=False)
print(f"\nUnique churn reasons saved to 'unique_churn_reasons_cluster_7043.csv'")
```

## Summary

The notebook performs the following tasks:
1. Loads both CSV files
2. Joins them by row number (index-based concatenation)
3. Saves the joined dataset as `joined_telco_retention.csv`
4. Filters the joined data where `L1_cluster_id = 7044`
5. Extracts unique `Churn Reason` values with counts
6. Saves the unique churn reasons to `unique_churn_reasons_cluster_7044.csv`
