# Preprocess Telco Customer Churn Dataset

This notebook loads the Telco customer churn dataset and removes unnecessary columns (geographic and identifier columns).

```python
import pandas as pd
import os

# Define file paths
source_file = r"C:\Users\weezh\OneDrive\Desktop\pyhercules\dataset\Telco customer churn IBM dataset\Telco_customer_churn.csv"
output_dir = r"C:\Users\weezh\OneDrive\Desktop\pyhercules\dataset\3rd exploration mdl implimentation"
output_file = os.path.join(output_dir, "Telco_customer_churn_cleaned.csv")

# Load the dataset
print("Loading dataset...")
df = pd.read_csv(source_file)
print(f"Original shape: {df.shape}")
print(f"Original columns: {list(df.columns)}")
```

```python
# Define columns to drop
columns_to_drop = [
    'CustomerID', 
    'Count', 
    'Country', 
    'State', 
    'City', 
    'Zip Code', 
    'Lat Long', 
    'Latitude', 
    'Longitude'
]

# Check which columns exist in the dataset
existing_columns_to_drop = [col for col in columns_to_drop if col in df.columns]
missing_columns = [col for col in columns_to_drop if col not in df.columns]

print(f"\nColumns to drop (found in dataset): {existing_columns_to_drop}")
if missing_columns:
    print(f"Columns not found in dataset: {missing_columns}")

# Drop the columns
df_cleaned = df.drop(columns=existing_columns_to_drop)
print(f"\nCleaned shape: {df_cleaned.shape}")
print(f"Remaining columns: {list(df_cleaned.columns)}")
```

```python
# Save the cleaned dataset
print(f"\nSaving cleaned dataset to: {output_file}")
df_cleaned.to_csv(output_file, index=False)
print(f"✓ Dataset saved successfully!")
print(f"  - Dropped {len(existing_columns_to_drop)} columns")
print(f"  - Retained {df_cleaned.shape[1]} columns")
print(f"  - Total rows: {df_cleaned.shape[0]}")
```
