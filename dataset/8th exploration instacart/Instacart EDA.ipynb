## Dataset Relationships

**Data Flow:**
```
USER (user_id)
  └─> ORDERS (order_id, user_id, order_number, order_dow, order_hour_of_day, days_since_prior_order)
       └─> ORDER_PRODUCTS (order_id, product_id, add_to_cart_order, reordered)
            └─> PRODUCTS (product_id, product_name, aisle_id, department_id)
                 ├─> AISLES (aisle_id, aisle)
                 └─> DEPARTMENTS (department_id, department)
```

**Key Relationships:**
- `orders.user_id` → Links to unique customers (~206k users)
- `order_products_*.order_id` → Links to `orders.order_id`
- `order_products_*.product_id` → Links to `products.product_id`
- `products.aisle_id` → Links to `aisles.aisle_id`
- `products.department_id` → Links to `departments.department_id`

**Train/Prior Split:**
- `order_products__prior`: Historical orders for training (32.4M items)
- `order_products__train`: Target orders for prediction (1.3M items)
- `orders.eval_set`: 'prior', 'train', or 'test' designation

## Feature Engineering - Building User-Level Matrix

````python
import numpy as np
from tqdm import tqdm

# Step 1: Merge order_products with orders (use only 'prior' orders for feature extraction)
print("Step 1: Merging order_products with orders...")
prior_orders = orders[orders['eval_set'] == 'prior']
order_products_merged = order_products_prior.merge(prior_orders, on='order_id', how='left')
print(f"Merged shape: {order_products_merged.shape}")

# Step 2: Merge with products to get product details
print("\nStep 2: Merging with products...")
order_products_full = order_products_merged.merge(products, on='product_id', how='left')
print(f"Full merged shape: {order_products_full.shape}")

# Step 3: Merge with departments and aisles
print("\nStep 3: Merging with departments and aisles...")
order_products_full = order_products_full.merge(departments, on='department_id', how='left')
order_products_full = order_products_full.merge(aisles, on='aisle_id', how='left')
print(f"Complete data shape: {order_products_full.shape}")

print("\n✓ Data merging complete!")
print(f"Columns: {list(order_products_full.columns)}")
````

### Feature Set 1: Purchase Behavior

````python
print("Calculating purchase behavior features...")

# Calculate per user
user_features = pd.DataFrame()

# 1. Total number of orders
user_features['total_orders'] = prior_orders.groupby('user_id')['order_id'].nunique()

# 2. Average days between orders
user_features['avg_days_between_orders'] = prior_orders.groupby('user_id')['days_since_prior_order'].mean()

# 3. Average basket size (items per order)
basket_sizes = order_products_full.groupby(['user_id', 'order_id']).size().reset_index(name='basket_size')
user_features['avg_basket_size'] = basket_sizes.groupby('user_id')['basket_size'].mean()

# 4. Reorder rate
user_features['reorder_rate'] = order_products_full.groupby('user_id')['reordered'].mean()

# 5. Total products purchased
user_features['total_products_purchased'] = order_products_full.groupby('user_id')['product_id'].count()

# 6. Unique products purchased
user_features['unique_products'] = order_products_full.groupby('user_id')['product_id'].nunique()

print(f"✓ Purchase behavior features: {user_features.shape}")
user_features.head()
````

### Feature Set 2: Shopping Patterns

````python
print("Calculating shopping pattern features...")

# 1. Most common shopping day (mode)
user_features['preferred_day_of_week'] = prior_orders.groupby('user_id')['order_dow'].agg(lambda x: x.mode()[0] if len(x.mode()) > 0 else x.mean())

# 2. Most common shopping hour (mode)
user_features['preferred_hour_of_day'] = prior_orders.groupby('user_id')['order_hour_of_day'].agg(lambda x: x.mode()[0] if len(x.mode()) > 0 else x.mean())

# 3. Shopping day consistency (std deviation)
user_features['day_consistency'] = prior_orders.groupby('user_id')['order_dow'].std().fillna(0)

# 4. Shopping hour consistency (std deviation)
user_features['hour_consistency'] = prior_orders.groupby('user_id')['order_hour_of_day'].std().fillna(0)

# 5. Order frequency consistency (std of days between orders)
user_features['frequency_consistency'] = prior_orders.groupby('user_id')['days_since_prior_order'].std().fillna(0)

print(f"✓ Shopping pattern features: {user_features.shape}")
user_features.head()
````

### Feature Set 3: Product Preferences (Department Distribution)

````python
print("Calculating department preference features...")

# Calculate department purchase counts per user
dept_counts = order_products_full.groupby(['user_id', 'department']).size().reset_index(name='count')

# Pivot to get department columns
dept_pivot = dept_counts.pivot(index='user_id', columns='department', values='count').fillna(0)

# Normalize to percentages (each row sums to 1)
dept_pct = dept_pivot.div(dept_pivot.sum(axis=1), axis=0)

# Add prefix to column names
dept_pct.columns = [f'dept_pct_{col.replace(" ", "_")}' for col in dept_pct.columns]

print(f"✓ Department features: {dept_pct.shape}")
print(f"   Departments: {list(dept_pct.columns)[:5]}...")  # Show first 5
dept_pct.head()
````

### Feature Set 4: Product Diversity

````python
print("Calculating product diversity features...")

# 1. Unique aisles shopped
user_features['unique_aisles'] = order_products_full.groupby('user_id')['aisle_id'].nunique()

# 2. Unique departments shopped
user_features['unique_departments'] = order_products_full.groupby('user_id')['department_id'].nunique()

# 3. Product exploration rate (unique products / total products purchased)
user_features['product_exploration_rate'] = user_features['unique_products'] / user_features['total_products_purchased']

# 4. Average products per aisle (diversity within aisles)
user_features['avg_products_per_aisle'] = user_features['unique_products'] / user_features['unique_aisles']

print(f"✓ Product diversity features: {user_features.shape}")
user_features.head()
````

### Combine All Features & Save

````python
print("Combining all features...")

# Merge department percentages with user features
user_features_final = user_features.join(dept_pct, how='left')

# Fill any NaN values (users might not have purchased from all departments)
user_features_final = user_features_final.fillna(0)

# Reset index to make user_id a column
user_features_final = user_features_final.reset_index()

print(f"\n{'='*80}")
print(f"FINAL FEATURE MATRIX")
print(f"{'='*80}")
print(f"Shape: {user_features_final.shape}")
print(f"Users: {len(user_features_final):,}")
print(f"Features: {len(user_features_final.columns)-1}")  # -1 for user_id
print(f"\nFeature columns:")
print(list(user_features_final.columns))

print(f"\n✓ Feature engineering complete!")
user_features_final.head(10)
````
